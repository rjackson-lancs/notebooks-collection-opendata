# Heavy ion jet quenching with ***ATLAS*** Open Data
### Book 1: MC datasets
The formation of the quark-gluon plasma (QGP) in heavy ion collisions causes energy strong jet energy loss  due to traversal through the hot, dense, and coloured medium.
This phonemenon was first observed in the ***ATLAS*** detector in 2010 [[arXiv:1011.6182v2](https://arxiv.org/abs/1011.6182)] through centrality-dependant dijet asymmetry in lead-lead collisions. This is the first of two notebooks which will replicate the results of this analysis with the heavy ion hard probes release of ***ATLAS*** Open Data.
